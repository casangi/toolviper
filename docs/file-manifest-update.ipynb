```python
import pathlib

import toolviper
```

```python
def make_random_file(file):
    import random
    import subprocess

    pathlib.Path(file).mkdir(mode=0o777, parents=False, exist_ok=True)
    with open(pathlib.Path(file).joinpath("data.bin"), "wb") as handle:
        handle.write(random.randbytes(1024))

    subprocess.run(["zip", "-r", f"{file}.zip", file])
```

```python
path = pathlib.Path().cwd()

toolviper.utils.data.update(path=str(path))
```

```python
_json = toolviper.utils.tools.open_json(str(path / "file.download.json"))
toolviper.utils.display.DataDict.html(_json["metadata"])
```
